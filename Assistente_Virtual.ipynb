```python
# Instalar módulos necessários no Google Colab
!pip install SpeechRecognition wikipedia geopy requests gtts pydub

import speech_recognition as sr
import webbrowser
import wikipedia
import requests
import os
from gtts import gTTS
from IPython.display import Audio
from geopy.geocoders import Nominatim

# Função para converter texto em fala usando gTTS
def falar(texto):
    tts = gTTS(text=texto, lang='pt')
    filename = "voz.mp3"
    tts.save(filename)
    return Audio(filename, autoplay=True)

# Verifica conexão com a internet
def verificar_conectividade():
    try:
        requests.get("https://www.google.com", timeout=5)
        return True
    except requests.ConnectionError:
        return False

# Captura entrada do usuário via texto (já que áudio não funciona no Colab)
def ouvir():
    comando = input("Digite seu comando: ").lower()
    return comando

# Pesquisa na Wikipedia
def pesquisar_wikipedia(termo):
    if not verificar_conectividade():
        return falar("Sem conexão com a internet.")
    try:
        resumo = wikipedia.summary(termo, sentences=2, auto_suggest=False)
        print(resumo)
        return falar(resumo)
    except wikipedia.exceptions.DisambiguationError:
        return falar("Muitos resultados encontrados, seja mais específico.")
    except wikipedia.exceptions.PageError:
        return falar("Não encontrei informações sobre isso.")

# Abre o YouTube para um termo específico
def abrir_youtube(termo):
    if not verificar_conectividade():
        return falar("Sem conexão com a internet.")
    url = f"https://www.youtube.com/results?search_query={termo.replace(' ', '+')}"
    webbrowser.open(url)
    return falar(f"Abrindo YouTube para {termo}")

# Busca localização no Google Maps
def buscar_localizacao(termo):
    if not verificar_conectividade():
        return falar("Sem conexão com a internet.")
    geolocator = Nominatim(user_agent="assistente_virtual")
    location = geolocator.geocode(termo)
    if location:
        url = f"https://www.google.com/maps/search/{termo.replace(' ', '+')}"
        webbrowser.open(url)
        return falar(f"Aqui está a localização de {termo}")
    else:
        return falar("Não consegui encontrar esse local.")

# Identifica e executa o comando digitado
def executar_comando():
    comando = ouvir()
    if "pesquisar" in comando:
        termo = comando.replace("pesquisar", "").strip()
        pesquisar_wikipedia(termo)
    elif "youtube" in comando:
        termo = comando.replace("youtube", "").strip()
        abrir_youtube(termo)
    elif "onde fica" in comando:
        termo = comando.replace("onde fica", "").strip()
        buscar_localizacao(termo)
    elif "sair" in comando or "fechar assistente" in comando:
        falar("Até logo!")
        exit()
    else:
        falar("Desculpe, não reconheci o comando.")

# Executa o assistente
if __name__ == "__main__":
    falar("Olá! Eu sou seu assistente virtual. Como posso ajudar?")
    while True:
        executar_comando()
```
